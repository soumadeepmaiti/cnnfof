```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_pickle('/u/masou/My_Workspace/Git/haloes/Aletheia_halo_finder/cnnfof/data/bin/0003.bin')

field = df[df['Host'] == 0]
halo  = df[df['Host'] == 1]

fig, ax = plt.subplots(figsize=(10, 10))
ax.scatter(field['Px'], field['Py'], s=1, c='steelblue', alpha=0.2, label='field')
ax.scatter(halo['Px'], halo['Py'], s=3, c='limegreen', alpha=1.0, label='halo-parts')
ax.scatter(halo['Hx'],  halo['Hy'],  s=50, c='tomato', marker='x',   alpha=1, label='halo-centre')
ax.set(xlabel='X [Mpc]', ylabel='Y [Mpc]')
# ax.legend(markerscale=6)
plt.tight_layout()
plt.show()

print(f"Particles: {len(df):,}  |  halo: {len(halo):,}  |  field: {len(field):,}")
```
